```python
z=np.array([[1,2,3,4],[3,2,1,5],[9,8,7,6],[4,5,3,2]])
print(z)
print(z[0])
print(z[2])
print(z[1][1])
print(z[1,1])
print(z[0:,1]) #0 th row started till end, then comma 1st column
print(z[0:,1:3]) #prints 2nd and 3rd column
print(z[2,1:3]) # to print 8 7 in 3rd row
print(z[1:3,1:3])
```

```python
#Zero Matrix
z=np.zeros([3,3],dtype=int) #gives a 3 by 3 matrix with integer entries
print(z)

#One Matrix
z=np.ones([3,3],dtype=complex) #gives a one matrix with complex entries
print(z)
```

```python
# Create two 2x2 matrices and store it in two variables.
# Find sum, difference, product
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print(a)
print(b)
print(a+b)
print(a-b)
print(a*b) # not matrix multiplication
print(a@b) #matrix multiplication
#OR#
print(a.dot(b))
```

```python
# Full Matrix
z=np.full([3,3],4) # 3by3Scalar matrix with scalar 4
print(z)
```

```python
# Create a 4x4 matrix with value 9
A=np.full([4,4],9)
print(A)
```

```python
# Identity Matrix
I=np.identity(3)
print(I)
# OR #
I=np.eye(4,dtype=int)
print(I)
```

```python
#arange
z=np.arange(1,10)
print(z)
```

```python
# Create a 4x3 matrix
A=np.ones([4,3],dtype=int)
print(A)
#we can reshape the matrix(just mn should be same for m and n)
A=np.ones([4,3],dtype=int).reshape(6,2)
print(A)
```

```python
# Create a 3x3 matrix with arange and reshape
A=np.arange(1,10).reshape(3,3) #here array 1 to 9 (1x9 matrix) taken
print(A)                       #and then reshaped to a 3x3 matrix
```

```python
# Create a 4x4 matrix with float values.
z=np.array([[1.2,2.2,3.2,4.2],[2.5,4.5,3.5,7.5],[3.1,2.1,1.1,4.1],[5.3,1.3,2.3,6.3]])
print(z)
#floor
print(np.floor(z))
#ceil
print(np.ceil(z))
#round
print(np.round(z)) #rounded to nearest even number
```

```python
z=np.array([[1.2,2.2,3.2,4.2],[2.5,4.5,3.5,7.5],[3.1,2.1,1.1,4.1]])
print(z.max())
print(z.min())
print(z.sum())
```

```python
# to find max in each row
z=np.array([[1.2,2.2,3.2,4.2],[2.5,4.5,3.5,7.5],[3.1,2.1,1.1,4.1]])
print(z.max(axis=1)) #in numpy axis=1 means row wise(pandas it means columnwise)
print(z.max(axis=0)) #in nmpy axis=0 column(in pandas its row)
#similarly we can use sum(axis=1) and sum(axis=0) to find max in rowwise and columnwise
print(np.sin(z))
print(np.cos(z))
print(np.sort(z))
print(np.sort(z,axis=0))
print(np.argsort(z,axis=1)) #index of each element of soretd z in original z
```
